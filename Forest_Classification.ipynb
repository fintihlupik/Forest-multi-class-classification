# ¿Qué es el dataset Covertype? 🌲
Cada fila representa una parcela de tierra en un bosque nacional de Colorado, EE.UU., con características como:

1. Elevación, inclinación, distancia a ríos o caminos...

2. Tipo de suelo (40 columnas one-hot)

3. Zona silvestre (4 columnas one-hot)

Y una columna objetivo: Cover_Type (1 a 7), que representa el tipo de vegetación principal (ej. pino, álamo, etc.)


-----------

## Descripción del dataset

The dataset includes four wilderness areas located in the Roosevelt National Forest in northern Colorado. The Roosevelt National Forest is a National Forest, located in north central Colorado. Each observation is a 30m x 30m patch.

- Elevation - Elevation in meters.
- Aspect - Aspect in degrees azimuth. La herramienta Aspecto (Aspect) identifica la dirección hacia la cual se orienta la pendiente descendente.
Los valores de cada celda en el ráster de salida indican la dirección de la brújula (rumbo) hacia la cual se orienta la superficie en ese punto.
Se mide en sentido horario en grados, desde 0° (norte) hasta 360° (norte otra vez), completando un círculo completo.
Las áreas planas que no tienen una dirección de pendiente descendente se les asigna un valor de -1.

  Aspecto (Aspect): Es la dirección hacia la cual una pendiente mira, como si el agua corriera cuesta abajo desde ese punto. Se usa en geografía, hidrología, agricultura, etc.

          0° y 360°: Ambos representan el norte.
          0° = norte
          90° = este
          180° = sur
          270° = oeste
          360° = norte otra vez

- Slope - Slope in degrees. Slope (pendiente): Es una medida de qué tan empinada o inclinada está una superficie. Se expresa comúnmente en grados o porcentaje.

- Horizontal_Distance_To_Hydrology - Horz Dist to nearest surface water features.
- Vertical_Distance_To_Hydrology - Vert Dist to nearest surface water features.
- Horizontal_Distance_To_Roadways - Horz Dist to nearest roadway.
- Hillshade_9am (0 to 255 index) - Hillshade index at 9am, summer solstice. Es una representación visual del terreno que simula sombras y luces como si el sol iluminara la superficie desde un ángulo específico. No representa elevación, sino intensidad de iluminación: 0 - sombra total, 255 - luz máxima.

- Hillshade_Noon (0 to 255 index) - Hillshade index at noon, summer solstice.

- Hillshade_3pm (0 to 255 index) - Hillshade index at 3pm, summer solstice.
- Horizontal_Distance_To_Fire_Points - Horz Dist to nearest wildfire ignition points.
- Wilderness_Area (4 binary columns, 0 = absence or 1 = presence) - Wilderness area designation.
- Soil_Type (40 binary columns, 0 = absence or 1 = presence) - Soil Type designation.
- Cover_Type (7 types, integers 1 to 7) - Forest Cover Type designation.


Seven Types of Forest Cover:

1. Spruce/Fir (Pícea/Abeto)
2. Lodgepole Pine. (Pino Lodgepole)
3. Ponderosa Pine. (Pino Ponderosa)
4. Cottonwood/Willow. (Álamo/Sauce)
5. Aspen. (Álamo Temblón)
6. Douglas-fir. (Abeto de Douglas)
7. Krummholz. (Bosque enano alpino, vegetación de alta montaña)

```python
!wget https://archive.ics.uci.edu/ml/machine-learning-databases/covtype/covtype.data.gz

import pandas as pd
import gzip

# Abrir y leer el archivo comprimido
with gzip.open('covtype.data.gz', 'rt') as f:
    df = pd.read_csv(f, header=None)

# Mostrar las primeras filas
df.head()
```

### Outliers

```python
non_binary_features = df.drop(columns=['Cover_Type']).loc[:, df.drop(columns=['Cover_Type']).nunique() > 2]
```

```python
def calculate_outliers_percentage(df):

    # Calcular los cuartiles y el IQR
    Q1 = df.quantile(0.25)
    Q3 = df.quantile(0.75)
    IQR = Q3 - Q1

    # Límites para detectar outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Contar los outliers por columna
    num_outliers = ((df < lower_bound) | (df > upper_bound)).sum()

    # Calcular el porcentaje de outliers
    pct_outliers = round(num_outliers / len(df) * 100, 4)

    return pct_outliers
```

```python
outliers = pd.DataFrame(calculate_outliers_percentage(non_binary_features), columns=['% Outliers']).sort_values(by='% Outliers', ascending=False)
outliers.index.name = 'Column Name'
outliers.reset_index(inplace=True)
outliers
```

Estos outliers son valores válidos ya que representan la variabilidad del dato, por lo que no se aconseja su eliminación. Dependiendo del modelo existen las opciones de transformarlos (p.ej. con log o winsorization) o tratarlos por separado.

### Cramers Correlations

```python
from scipy.stats import chi2_contingency

# Función para calcular Cramér's V
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

# Excluir Cover_Type y detectar columnas binarias
binary_cols = [col for col in df.columns if df[col].nunique() == 2]

# Calcular Cramér’s V con Cover_Type
cramers_results = {
    col: cramers_v(df['Cover_Type'], df[col]) for col in binary_cols
}

# Convertir a DataFrame
cramers_df = pd.DataFrame.from_dict(cramers_results, orient='index', columns=["Cramér's V"])
cramers_df = cramers_df.sort_values(by="Cramér's V", ascending=False)
cramers_df.reset_index(inplace=True)
cramers_df.rename(columns={'index': 'Variable'}, inplace=True)

# Mostrar resultado
#print(cramers_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# (Este bloque debe ejecutarse después de generar cramers_df como en el código anterior)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=cramers_df,
    y='Variable',
    x="Cramér's V",
    palette='viridis'
)

plt.title("Cramér's V entre Cover_Type y variables binarias", fontsize=14)
plt.xlabel("Cramér's V")
plt.ylabel("Variable Binaria")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Splitting

```python
X = df.drop('Cover_Type', axis=1)
y = df['Cover_Type']

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)
```

## Balance

```python
print((df.groupby('Cover_Type').Aspect.count() / len(df)).apply(lambda p: f"{p:.3%}"))
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
```

```python
# Comparación gráfica de la distribución de clases antes y después de SMOTE

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Antes de SMOTE
sns.countplot(x=y_train, ax=axes[0])
axes[0].set_title('Distribución de clases ANTES de SMOTE')
axes[0].set_xlabel('Tipo de cobertura forestal')
axes[0].set_ylabel('Cantidad')

# Después de SMOTE
sns.countplot(x=y_train_res, ax=axes[1])
axes[1].set_title('Distribución de clases DESPUÉS de SMOTE')
axes[1].set_xlabel('Tipo de cobertura forestal')
axes[1].set_ylabel('Cantidad')

plt.tight_layout()
plt.show()
```
